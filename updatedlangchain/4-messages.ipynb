## Messages
Messages are the fundamental unit of context for models in LangChain. They repressetn the input and the output of model, carrying both the content and metadata needed to represent the state of a conversation when interacting with an LLM. Messages are objects that contain:
1. Role - Identifies the message type (eg. system, user)
2. Content - Represent the actual content of the message (like text, images, audio, documents, etc)
3. Metadata - Optional fields such as response information, message IDs, and token usage

Langchain provides a standard message type that works across all model provides, ensuring consistent behavior regardless of the model being called.

```python
import os
from langchain.chat_models import init_chat_model
os.environ["GOOGLE_API_KEY"] = os.getenv("GOOGLE_API_KEY")

model = init_chat_model("google_genai:gemini-2.5-flash-lite")
```

```python
model.invoke("What is the capital of India? Answer in one word only.")
```

# Message Prompts
Alternatively, you can pass in a list of messages to the model by providing a list of message objects.
Message types
1. System message - Tells the model how t behave and provide context for interactions
2. Human Message - Represents user input and interatcions with the model
3. AI message - Responses generated by the model, including text conten, tool calls and metadata
4. Tool message - Represents the outputs of a tool call.

## System Message 
A System MEssage represent a initial set of instructions that primes the model's behaviour.

## Human Message
A humana message represents the message sent by the user invoking the model and what he is trying to ask from the model.

## AI Message
An ai message represents the response given by the model to the human message that was sent by the user earlier

## Tool Message
A tool message is the message that has been given back by the tool as a response on it's invocation.

```python
from langchain.messages import SystemMessage, HumanMessage, AIMessage

messages = [
    SystemMessage("You are an expert song writer like Karan Aujla"),
    HumanMessage("Write a song on desi culture and hardships faced by desi people in foreign countries."),
]

response = model.invoke(messages)
response.content
```

```python
## Detailed information to the LLM thuorgh System MEssage
from langchain.messages import SystemMessage, HumanMessage

system_msg = SystemMessage ("""
You are a senior Python developer with expertise in web frameworks.
Always provide code examples and explain your reasoning.
Be concise but thorough in your explanations.
""")

messages = [
    system_msg,
    HumanMessage("How do i create REST API in Python?")
]

response = model.invoke(messages)
print(response.content)
```
