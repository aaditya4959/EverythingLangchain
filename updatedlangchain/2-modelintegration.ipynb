### Models integration with various llm providers like openai, google gemini and groq.

```python
import os
from dotenv import load_dotenv
load_dotenv()

os.environ["OPENAI_API_KEY"] = os.getenv("OPENAI_API_KEY")
os.environ["GROQ_API_KEY"] = os.getenv("GROQ_API_KEY")
os.environ["GOOGLE_API_KEY"] = os.getenv("GOOGLE_API_KEY")
```

```python
from langchain.chat_models import init_chat_model
model = init_chat_model("gpt-4.1", temperature=0.7)
model
```

```python
### invoke the model
response = model.invoke("Hello how are you?")
response
```

```python
### Google gemini models
from langchain.chat_models import init_chat_model

model = init_chat_model("google_genai:gemini-2.5-flash-lite")
response = model.invoke("Why do parrots talk?")
response.content
```

```python
response.content
```

```python
### ChatOPenAI
## again llm limit lol.

from langchain_openai import ChatOpenAI
model = ChatOpenAI(model="gpt-4.1", temperature=0.7)
response = model.invoke("What is the capital of France?")
response.content
```

```python
### GROQ model integraion

from langchain.chat_models import init_chat_model

model = init_chat_model("groq:qwen/qwen3-32b")
response = model.invoke("What is the capital of France?")
response
```

```python
### GROQ model integration with specified library

from langchain_groq import ChatGroq
model = ChatGroq(model="qwen/qwen3-32b", temperature=0)
response = model.invoke("What is the capital of France?")
response.content
```
